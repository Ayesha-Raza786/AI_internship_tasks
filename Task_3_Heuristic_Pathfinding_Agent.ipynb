```python
import heapq
import time
import random
import matplotlib.pyplot as plt
import pandas as pd
from collections import deque

# -----------------------------
# 1. Grids
# -----------------------------

easy_grid = [
    ['S', '.', '.', '.', '.'],
    ['.', '.', '.', '.', '.'],
    ['.', '.', '.', '.', '.'],
    ['.', '.', '.', '.', '.'],
    ['.', '.', '.', '.', 'G']
]

medium_grid = [
    ['S', '.', '#', '.', '.'],
    ['.', '.', '#', '.', '.'],
    ['#', '.', '#', '.', '.'],
    ['.', '.', '.', '.', '#'],
    ['.', '#', '.', '.', 'G']
]

hard_grid = [
    ['S', '.', '#', '#', '.'],
    ['.', '.', '.', '#', '.'],
    ['#', '#', '.', '#', '.'],
    ['.', '.', '.', '.', '.'],
    ['#', '#', '#', '.', 'G']
]

custom_grid = [
    ['S', '.', '.', '#', '.'],
    ['.', '#', '.', '#', '.'],
    ['.', '#', '.', '.', '.'],
    ['.', '.', '.', '#', '.'],
    ['#', '.', '.', '.', 'G']
]

challenge_grid = [
    ['S', '#', '.', '.', '.'],
    ['.', '#', '.', '#', '.'],
    ['.', '#', '.', '#', '.'],
    ['.', '.', '.', '#', '.'],
    ['#', '#', '.', '.', 'G']
]

detour_grid = [
    ['S', '.', '#', '.', '#'],
    ['.', '.', '.', '#', '.'],
    ['.', '.', '#', '#', '.'],
    ['.', '#', '.', '.', '.'],
    ['.', '.', '.', '#', 'G']
]

mazes = {
    "Easy": easy_grid,
    "Medium": medium_grid,
    "Hard": hard_grid,
    "Custom": custom_grid,
    "Challenge": challenge_grid,
    "Detour": detour_grid
}

start = (0, 0)
goal = (4, 4)


# -----------------------------
# 2. Get valid neighbors
# -----------------------------

def get_neighbors(grid, node):
    rows = len(grid)
    cols = len(grid[0])
    row, col = node

    directions = [(-1, 0), (1, 0), (0, -1), (0, 1)]

    neighbors = []
    for dr, dc in directions:
        new_row = row + dr
        new_col = col + dc
        if 0 <= new_row < rows and 0 <= new_col < cols:
            if grid[new_row][new_col] != '#':
                neighbors.append((new_row, new_col))

    return neighbors


# -----------------------------
# 3. Manhattan heuristic
# -----------------------------

def heuristic(node, goal):
    return abs(node[0] - goal[0]) + abs(node[1] - goal[1])


# -----------------------------
# 4. A* Algorithm
# -----------------------------

def a_star(grid, start, goal):
    start_time = time.perf_counter()

    open_list = []
    heapq.heappush(open_list, (0, start))

    g_score = {start: 0}
    came_from = {}
    closed_set = set()
    order = []

    expanded_nodes = 0

    while open_list:
        current_f, current = heapq.heappop(open_list)

        if current in closed_set:
            continue
        closed_set.add(current)
        order.append(current)

        expanded_nodes += 1

        if current == goal:
            path = reconstruct_path(came_from, current)
            runtime = time.perf_counter() - start_time
            return path, expanded_nodes, runtime, order

        for neighbor in get_neighbors(grid, current):
            if neighbor in closed_set:
                continue

            new_g = g_score[current] + 1

            if neighbor not in g_score or new_g < g_score[neighbor]:
                g_score[neighbor] = new_g
                came_from[neighbor] = current
                h = heuristic(neighbor, goal)
                f = new_g + h
                heapq.heappush(open_list, (f, neighbor))

    runtime = time.perf_counter() - start_time
    return None, expanded_nodes, runtime, order


# -----------------------------
# 4b. Dijkstra (uniform cost search, no heuristic)
# -----------------------------

def dijkstra(grid, start, goal):
    start_time = time.perf_counter()

    open_list = []
    heapq.heappush(open_list, (0, start))

    distance = {start: 0}
    came_from = {}
    closed_set = set()
    expanded_nodes = 0

    while open_list:
        current_distance, current = heapq.heappop(open_list)

        if current in closed_set:
            continue
        closed_set.add(current)
        expanded_nodes += 1

        if current == goal:
            path = reconstruct_path(came_from, current)
            runtime = time.perf_counter() - start_time
            return path, expanded_nodes, runtime

        for neighbor in get_neighbors(grid, current):
            if neighbor in closed_set:
                continue

            new_distance = distance[current] + 1

            if neighbor not in distance or new_distance < distance[neighbor]:
                distance[neighbor] = new_distance
                came_from[neighbor] = current
                heapq.heappush(open_list, (new_distance, neighbor))

    runtime = time.perf_counter() - start_time
    return None, expanded_nodes, runtime


# -----------------------------
# 5. Reconstruct final path
# -----------------------------

def reconstruct_path(came_from, current):
    path = [current]
    while current in came_from:
        current = came_from[current]
        path.append(current)
    path.reverse()
    return path


# -----------------------------
# 6. Visualization
# -----------------------------

import os

OUTPUT_DIR = "pathfinding_logs"
os.makedirs(OUTPUT_DIR, exist_ok=True)


def visualize_grid(grid, path, start, goal, title="A* Pathfinding", order=None, save_path=None):
    rows = len(grid)
    cols = len(grid[0])

    plt.figure(figsize=(7, 7))

    for row in range(rows):
        for col in range(cols):
            if grid[row][col] == '#':
                plt.fill(
                    [col, col + 1, col + 1, col],
                    [rows - row - 1, rows - row - 1, rows - row, rows - row],
                    color='black'
                )

    if order:
        for i, (row, col) in enumerate(order):
            if (row, col) in (start, goal):
                continue
            plt.text(
                col + 0.5, rows - row - 0.5, str(i),
                ha='center', va='center', fontsize=8, color='gray'
            )

    if path:
        x = [col + 0.5 for row, col in path]
        y = [rows - row - 0.5 for row, col in path]
        plt.plot(x, y, marker='o', linewidth=3, color='red', zorder=5)

    plt.text(start[1] + 0.5, rows - start[0] - 0.5, 'S', ha='center', va='center', fontsize=15, color='green', weight='bold')
    plt.text(goal[1] + 0.5, rows - goal[0] - 0.5, 'G', ha='center', va='center', fontsize=15, color='blue', weight='bold')

    plt.xlim(0, cols)
    plt.ylim(0, rows)
    plt.xticks(range(cols + 1))
    plt.yticks(range(rows + 1))
    plt.grid(True)
    plt.title(title)

    if save_path:
        plt.savefig(save_path, dpi=150, bbox_inches='tight')
        print(f"Saved visualization: {save_path}")

    plt.show()


def explain_path(path, goal):
    print("Step | Position | g (cost so far) | h (est. remaining) | f (g+h)")
    for g, node in enumerate(path):
        h = heuristic(node, goal)
        f = g + h
        print(f"{g:>4} | {str(node):>8} | {g:>16} | {h:>19} | {f:>7}")


# -----------------------------
# 7. Run A* on every fixed maze and visualize each one
# -----------------------------

for name, maze in mazes.items():
    path, expanded_nodes, runtime, order = a_star(maze, start, goal)

    print("\nMaze:", name)

    if path:
        print("Path found:", True)
        print("Steps:", len(path) - 1)
        print("Expanded nodes:", expanded_nodes)
        print("Runtime:", runtime, "seconds")
        explain_path(path, goal)
    else:
        print("Path found:", False)
        print("Expanded nodes:", expanded_nodes)
        print("Runtime:", runtime, "seconds")

    visualize_grid(
        maze, path, start, goal,
        title=f"A* Pathfinding - {name}",
        order=order,
        save_path=os.path.join(OUTPUT_DIR, f"maze_{name.lower()}.png")
    )


# -----------------------------
# 8. A* vs Dijkstra on the same fixed mazes
# -----------------------------

print("\n" + "=" * 60)
print("A* vs DIJKSTRA COMPARISON")
print("=" * 60)

for name, maze in mazes.items():
    print(f"\n{name} Maze")
    print("-" * 40)

    a_path, a_expanded, a_runtime, _ = a_star(maze, start, goal)
    d_path, d_expanded, d_runtime = dijkstra(maze, start, goal)

    if a_path:
        print("A*:")
        print("  Steps:", len(a_path) - 1)
        print("  Expanded nodes:", a_expanded)
        print("  Runtime:", a_runtime)
    else:
        print("A*: No path found")

    if d_path:
        print("Dijkstra:")
        print("  Steps:", len(d_path) - 1)
        print("  Expanded nodes:", d_expanded)
        print("  Runtime:", d_runtime)
    else:
        print("Dijkstra: No path found")


# -----------------------------
# 9. Random maze generation + solvability check
# -----------------------------

def generate_random_grid(rows, cols, obstacle_probability):
    grid = []
    for row in range(rows):
        current_row = []
        for col in range(cols):
            if random.random() < obstacle_probability:
                current_row.append('#')
            else:
                current_row.append('.')
        grid.append(current_row)

    grid[0][0] = 'S'
    grid[rows - 1][cols - 1] = 'G'
    return grid


def has_path(grid, start, goal):
    queue = deque([start])
    visited = {start}

    while queue:
        current = queue.popleft()
        if current == goal:
            return True
        for neighbor in get_neighbors(grid, current):
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)

    return False


def generate_solvable_grid(rows, cols, obstacle_probability, max_attempts=1000):
    for _ in range(max_attempts):
        grid = generate_random_grid(rows, cols, obstacle_probability)
        if has_path(grid, (0, 0), (rows - 1, cols - 1)):
            return grid
    return None


# -----------------------------
# 10. Demo: A* vs Dijkstra on one random 10x10 maze
# -----------------------------

random_rows, random_cols = 10, 10
random_grid = generate_solvable_grid(random_rows, random_cols, 0.20)

print("\nRandom Maze:")
for row in random_grid:
    print(row)

# FIX: goal must match this grid's actual size, not the leftover (4,4)
# from the 5x5 mazes section above.
random_start = (0, 0)
random_goal = (random_rows - 1, random_cols - 1)

a_path, a_expanded, a_runtime, a_order = a_star(random_grid, random_start, random_goal)
d_path, d_expanded, d_runtime = dijkstra(random_grid, random_start, random_goal)

print("\n" + "=" * 50)
print("A* vs Dijkstra - Random 10x10 Maze")
print("=" * 50)

print("\nA*")
print("Steps:", len(a_path) - 1)
print("Expanded nodes:", a_expanded)
print("Runtime:", a_runtime)

print("\nDijkstra")
print("Steps:", len(d_path) - 1)
print("Expanded nodes:", d_expanded)
print("Runtime:", d_runtime)

visualize_grid(
    random_grid,
    a_path,
    random_start,
    random_goal,
    title="A* on Random 10x10 Maze",
    order=a_order,
    save_path=os.path.join(OUTPUT_DIR, "random_maze_10x10.png")
)


# -----------------------------
# 11. Benchmark across sizes and obstacle densities
# -----------------------------

results = []

sizes = [10, 20, 30]
obstacle_probabilities = [0.10, 0.20, 0.30]

for size in sizes:
    for obstacle_probability in obstacle_probabilities:
        grid = generate_solvable_grid(size, size, obstacle_probability)

        bench_start = (0, 0)
        bench_goal = (size - 1, size - 1)

        a_path, a_expanded, a_runtime, _ = a_star(grid, bench_start, bench_goal)
        d_path, d_expanded, d_runtime = dijkstra(grid, bench_start, bench_goal)

        results.append({
            "Grid Size": size,
            "Grid Label": f"{size}x{size}",
            "Obstacle Probability": obstacle_probability,
            "Algorithm": "A*",
            "Steps": len(a_path) - 1,
            "Expanded Nodes": a_expanded,
            "Runtime (ms)": a_runtime * 1000
        })

        results.append({
            "Grid Size": size,
            "Grid Label": f"{size}x{size}",
            "Obstacle Probability": obstacle_probability,
            "Algorithm": "Dijkstra",
            "Steps": len(d_path) - 1,
            "Expanded Nodes": d_expanded,
            "Runtime (ms)": d_runtime * 1000
        })

results_df = pd.DataFrame(results)
print("\n", results_df)

# Export the raw metrics table as a log file
results_csv_path = os.path.join(OUTPUT_DIR, "benchmark_results.csv")
results_df.to_csv(results_csv_path, index=False)
print(f"Saved metrics log: {results_csv_path}")

# FIX: one line per (Algorithm, Obstacle Probability) combo, instead of
# mixing all 3 densities into a single line per algorithm. This removes
# the zig-zag/jagged appearance, since each line now has exactly one
# point per grid size, in a clean monotonic order.

plt.figure(figsize=(10, 6))
for algorithm in ["A*", "Dijkstra"]:
    for obstacle_probability in obstacle_probabilities:
        data = results_df[
            (results_df["Algorithm"] == algorithm) &
            (results_df["Obstacle Probability"] == obstacle_probability)
        ].sort_values("Grid Size")

        plt.plot(
            data["Grid Label"],
            data["Expanded Nodes"],
            marker="o",
            label=f"{algorithm} (p={obstacle_probability})"
        )

plt.xlabel("Grid Size")
plt.ylabel("Expanded Nodes")
plt.title("A* vs Dijkstra - Expanded Nodes")
plt.legend()
plt.grid(True)
plt.savefig(os.path.join(OUTPUT_DIR, "expanded_nodes_comparison.png"), dpi=150, bbox_inches='tight')
print(f"Saved chart: {os.path.join(OUTPUT_DIR, 'expanded_nodes_comparison.png')}")
plt.show()

plt.figure(figsize=(10, 6))
for algorithm in ["A*", "Dijkstra"]:
    for obstacle_probability in obstacle_probabilities:
        data = results_df[
            (results_df["Algorithm"] == algorithm) &
            (results_df["Obstacle Probability"] == obstacle_probability)
        ].sort_values("Grid Size")

        plt.plot(
            data["Grid Label"],
            data["Runtime (ms)"],
            marker="o",
            label=f"{algorithm} (p={obstacle_probability})"
        )

plt.xlabel("Grid Size")
plt.ylabel("Runtime (ms)")
plt.title("A* vs Dijkstra - Runtime")
plt.legend()
plt.grid(True)
plt.savefig(os.path.join(OUTPUT_DIR, "runtime_comparison.png"), dpi=150, bbox_inches='tight')
print(f"Saved chart: {os.path.join(OUTPUT_DIR, 'runtime_comparison.png')}")
plt.show()
```
